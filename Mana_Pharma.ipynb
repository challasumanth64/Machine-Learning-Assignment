```python
import pandas as pd

# Load the dataset from local file
df = pd.read_csv("CSV.csv")

# Display the first few rows
df.head()
```

```python
# Check column names and data types
print(df.info())

# Check for missing values
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Drop missing values if any
df_cleaned = df.dropna()

# Optional: rename columns for ease
df_cleaned.rename(columns={
    "CID": "Compound_ID",
    "MolecularWeight": "MolWeight",
    "XLogP": "LogP"
}, inplace=True)

df_cleaned.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))
sns.scatterplot(data=df_cleaned, x="MolWeight", y="LogP", hue="Compound_ID", palette="Set1")
plt.title("Molecular Weight vs LogP")
plt.xlabel("Molecular Weight")
plt.ylabel("LogP")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x='Compound_ID', y='LogP', data=df_cleaned, palette="Set2")
plt.title("LogP Value of Compounds")
plt.xlabel("Compound ID")
plt.ylabel("LogP")
plt.grid(True)
plt.show()
```
